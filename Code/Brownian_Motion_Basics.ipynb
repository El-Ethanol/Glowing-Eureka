# Brownian Motion 



Brownian motion is a random process that has application in engineering, physics, and finance. It is a gaussian random precess, it models the motion of particles suspended in a fluid, percentage changes in stock prices and integrated white noise. 

## Symmetric Random Walk

Toss a fair coint, with probability $p$ of heads and $q=1-p$ of tails. The succesive outcomes of the tosses are giben by:

$\omega = \omega_1\omega_2...$.

$\omega$ is an infinite sequence of toses.

Let 

\begin{equation}
X_n(\omega) = \begin{cases}
\sqrt{\delta} & \text{if } \omega_n = H \\
-\sqrt{\delta} & \text{if } \omega_n = T
\end{cases}
\end{equation}

and define $M_0 = 0$ and 

$M_k = \sum_{n=1}^{k} X_n$ for $k \geq 1$. Then $M_k$ is the position of the particle after $k$ tosses. A symmetric random walk. Note that $X_n$ are independent and identically distributed random variables with $E[X_n] = 0$ and $Var(X_n) = \delta$.


Since $M_k = \sum_{n=1}^{k} X_n$, we have $E[M_k] = 0$ and $Var(M_k) = Var(\sum_{n=1}^k X_n) = n Var(X_i) = k \delta$. 

If we assume, we divide the time interval $[0,T]$ into $N$ equal subintervals of length $\delta = T/N$, then we can define a continuous-time process $M(t)$ by setting $M(t) = M_k$ for $t \in [k\delta, (k+1)\delta)$. As $N \to \infty$, we have $\delta \to 0$ and $M(t)$ will become a normal random variable $\mathcal{N}(0,t)$. The process $M(t)$ is called a Brownian motion or standard Wiener process. 

## Formal definition of Brownian Motion

**Def.** (Brownian Motion): 

A Gaussian random process $\{B(t), t \geq 0\}$ is called a Brownian motion if it satisfies the following properties:

1. $B(0) = 0$ almost surely.
2. For $0 \leq s < t$, the increment $B(t) - B(s) \sim \mathcal{N}(0, t-s)$.
3. $B(t)$ has independent increments, i.e., for $0 \leq t_1 < t_2 < ... < t_n$, the random variables $B(t_2) - B(t_1), B(t_3) - B(t_2), ..., B(t_n) - B(t_{n-1})$ are independent. 
4. $B(t)$ has continuous paths, i.e., $t \mapsto B(t)$ is continuous almost surely.


A more general porcess is obtained if we define $X(t)=\mu+\sigma B(t)$. In this case, $B(t)$ is called a Brownian motion with drift $\mu$ and variance $\sigma^2$.
