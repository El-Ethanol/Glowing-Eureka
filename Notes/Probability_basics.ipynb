# Probability Basics

**Def**. *(Sample Space and Event)*: The sample space $S$ of an experiment is the set of all possible outcomes of the experiment. An event $A$ is a subset of the sample space $S$. We say that $A$ **occurs** if the outcome of the experiment is in $A$.

**Ex.** Consider the experiment of 10 coining flips. Codifing heads as $1$ and tails as $0$, the sample space $S$ is given by:

$$ S=\{(s_1, s_2, \ldots, s_{10}) : s_i \in \{0, 1\} \text{ for } i = 1, 2, \ldots, 10\} $$

Let $A_i$ be the event that the i-th flip is heads, then $A_i$ can be represented as:

$$ A_i = \{(s_1, s_2, s_3, \ldots, s_{10}) : s_j \in \{0, 1\} \text{ for } j = 1, 2, 3, \ldots, 10\} $$

It is clear that $A_i$ is a subset of $S$, and it contains all outcomes where the i-th flip is heads.

The event $B$ where there is at least one head can be represented as:

$$ B = \bigcup_{i=1}^{10} A_i $$

The event $C$ where all flips are heads, 

$$ C = \bigcap_{i=1}^{10} A_i $$

The event $D$ where there are at least two consecutive heads can be represented as:

$$ D = \bigcup_{i=1}^{9} (A_i \cap A_{i+1}) $$

**Def** _(Naive Probability)_: The probability of an event $A$ is defined as the ratio of the number of favorable outcomes to the total number of outcomes in the sample space. 

\begin{equation}
P(A) = \frac{|A|}{|S|}
\end{equation}

This definitions assumes that all outcomes in the sample space are equally likely. This is valid when there is symmetry in the experiment, such as flipping a fair coin or rolling a fair dice, or as a null model. 

To get the naive probability, we need to be able to count the number of outcomes in the sample space and the number of favorable outcomes. This can be done using combinatorial methods, but there is the _Multiplication Principle_ which is a more general method for counting outcomes.

**Theorem** _(Multiplication Principle)_: Consider a compound experiment consisting of two sub-experiments, $A$ and $B$. Suppose that the Experiment $A$ has $a$ possible outcomes, and for each outcome of $A$, the experiment $B$ has $b$ possible outcomes. Then the compound experiment has $a \cdot b$ possible outcomes.

**Theorem** _(Sampling with replacement)_: Consider $n$ objects and making $k$ choices from them, one at a time with replacement (i.e. choosing an object does not remove it from the set of objects). Then there are $n^k$ possible outcomes.

**Theorem** _(Sampling without replacement)_: Consider $n$ objects and making $k$ choices from them, one at a time without replacement (i.e. choosing an object does remove it from the set of objects). Then there are $n!/(n-k)!$  for $k\leq n$ possible outcomes.

But if we are able to count each possibility exactly $c$ times, then we can divide the total number of outcomes by $c$ to get the correct count. We are adjusting the overcounting. For doing this there is a extremely useful tool _the binomial coefficient_ whcih counts the numer of subsets of a certain seiz for a set, such as the number of waus to choose $k$ objects from a set of $n$ objects. 

**Def.** _(Binomial Coefficient)_: The binomial coefficient $\binom{n}{k}$ is the number of ways to choose $k$ objects from a set of $n$ objects. It is given by: 

\begin{equation}
\binom{n}{k} = \frac{n!}{k!(n-k)!}
\end{equation}

Now, if we want to have a more general definition of probability, we can define a Probability Space. 

**Def.** _(Probability Space)_: A probability space is a triple $(\Omega, \mathcal{F}, P)$ where: 

- $\Omega$ is the sample space, which is the set of all possible outcomes of an experiment.
- $\mathcal{F}$ is a $\sigma$-algebra of subsets of $\Omega$, which is a collection of events that we can assign probabilities to. It must satisfy the following properties: 
  1. $\Omega \in \mathcal{F}$ (the sample space is an event)
  2. If $A \in \mathcal{F}$, then $A^c \in \mathcal{F}$ (the complement of an event is also an event)
  3. If $A_1, A_2, A_3, \ldots \in \mathcal{F}$, then $\bigcup_{i=1}^{\infty} A_i \in \mathcal{F}$ (the countable union of events is also an event)
- $P$ is a probability measure, which is a function that assigns a probability to each event in $\mathcal{F}$. It must satisfy the following properties: 
  1. $P(A) \geq 0$ for all $A \in \mathcal{F}$ (probabilities are non-negative)
  2. $P(\Omega) = 1$ (the probability of the sample space is 1)
  3. If $A_1, A_2, A_3, \ldots \in \mathcal{F}$ are disjoint events (i.e., $A_i \cap A_j = \emptyset$ for all $i \neq j$), then $P\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} P(A_i)$ (the probability of the union of disjoint events is the sum of their probabilities)


This axioms doesn't tell us how probabilities should be interpreted. There are at least two interpretations of probability: the frequentist interpretation and the Bayesian interpretation.

- Frequentist interpretation: Probability represents a long-run frequency over a large number of repetitions of an experiment: if we say a coin has probability 1/2 of Heads, that means the coin would land Heads 50% of the time if we tossed it over and over and over.
- Bayesian interpretation: Probability represents a degree of belief in an event. For example, if we say that the probability of rain tomorrow is 0.7, that means we believe there is a 70% chance of rain tomorrow based on our current knowledge and information.
